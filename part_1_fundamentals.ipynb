### 10.3
Write a function that validates a username according to the following rules:
- Must be between 3 and 20 characters long
- Can only contain letters, numbers, and underscores
- Must start with a letter
Return True if the username is valid, False otherwise.

Example: print(validate_username("john_doe123")) # Should print True
Example: print(validate_username("123john")) # Should print False


````python
def validate_username(username):
    # Check that the username is between 3 and 20 characters.
    if len(username) < 3 or len(username) > 20:
        return False

    # Check that the username starts with a letter.
    if not username[0].isalpha():
        return False

    # Check that every character is a letter, number, or underscore.
    if not username.replace("_", "").isalnum():
        return False

    return True


print(validate_username("john_doe123"))  # Should print True
print(validate_username("123john"))      # Should print False
````

### 10.4
Write a function that performs a basic validation of an email address.
It can be a fairly simple function, or you could try to look up how it's done in real life (copy-pasting is fine - as long as you understand it)
- If it's a valid email address, return True
- If it's an invalid email address, return False

Call the function and show me how I could use it


````python
def validate_email(email):
    # Check that there is exactly one @ symbol.
    if email.count("@") != 1:
        return False

    # Split the email into username and domain.
    username, domain = email.split("@")

    # Check that both parts contain something.
    if not username or not domain:
        return False

    # Check that the domain contains a dot.
    if "." not in domain:
        return False

    return True


# Test the function
print(validate_email("john@example.com"))  # True
print(validate_email("johnexample.com"))   # False
````

### Task 16
Write a function `describe_temperature(temp)` that returns:
- `"freezing"` if the temperature is below 0
- `"cold"` if it is between 0 and 14
- `"mild"` if it is between 15 and 24
- `"warm"` if it is 25 or above
Use `if`, `elif`, and `else`.
Call the function a few times to show that it works.


````python
def describe_temperature(temp):
    # Check the temperature and return the correct description.
    if temp < 0:
        return "freezing"
    elif temp <= 14:
        return "cold"
    elif temp <= 24:
        return "mild"
    else:
        return "warm"


# Call the function a few times to show that it works.
print(describe_temperature(-5))  # freezing
print(describe_temperature(10))  # cold
print(describe_temperature(20))  # mild
print(describe_temperature(30))  # warm
````

````python
def return_first_negative_number(l):
    # Go through each number in the list.
    for number in l:
        # Check if the number is negative.
        if number < 0:
            return number


numbers_list = [4, 1, 34, 42, -20, 21, 49, -48, 40, 3, -21, 10, 37, 36]

print(return_first_negative_number(numbers_list))  # Should print -20
````

### Task 22

You have a list consisting of dictionaries. Each dictionary represents data for an individual student in a system.

You want to create a new list that only consists of students where "Passed" is True, how do you do it?

**Tip** Utilize lists, loops and if-statements.

**Answer with code**


````python
students = [
    {'name': 'John', 'age': 18, 'passed': True, 'major': 'computer science', 'GPA': 3.8},
    {'name': 'Jane', 'age': 19, 'passed': False, 'major': 'biology', 'GPA': 3.2},
    {'name': 'Bob', 'age': 20, 'passed': True, 'major': 'engineering', 'GPA': 3.5},
    {'name': 'Alice', 'age': 21, 'passed': False, 'major': 'history', 'GPA': 2.9},
    {'name': 'David', 'age': 22, 'passed': True, 'major': 'business', 'GPA': 3.6}
]

# Create an empty list for students who passed.
passed_students = []

# Loop through each student.
for student in students:
    # If the student passed, add them to the new list.
    if student['passed']:
        passed_students.append(student)

print(passed_students)
````
